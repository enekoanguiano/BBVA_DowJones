#### Pasos a seguir

Analizar los componentes de la serie y tratarla como es debida para convertirla en estacionaria.

Si estacionalidad -> descomposición estacional

Si no estacionalidad -> tratar tendencia

Si tras tratar la serie existe heterocedasticidad -> Modelizar la volatilidad (ARCH o GARCH)

En caso de modelizar la volatilidad, se han mirado las autocorrelaciones y autocorrelaciones parciales sobre los rendimientos del activo al cuadrado. Sobre los resultados mostrados en las gráficas ACF y PACF se han decidido los modelos a probar, así como qué parámetros utilizar para ajustar los ARCH y GARCH. En caso de obtener un caso donde no hay modelos válidos, pero un GARCH obtiene todos los parámetros significativos con excepción del parametro mu para la media, se ha optado por probar a ajustar un EGARCH.


Para admitir un modelo, debe de cumplir con los siguientes puntos:
- Parámetros de modelo significativos
- El residuo post-modelo es homocedástico (ya no queda volatilidad)

```python
def test_estacionariedad(series, lags=None):
    series = series.dropna()
    
    resultados = {}
    
    # 1. ADF
    adf_stat, adf_p, _, _, _, _ = adfuller(series, autolag='AIC')
    resultados['ADF'] = {
        'Estadístico': adf_stat,
        'p-value': adf_p,
        'Estacionariedad (0.05)': adf_p < 0.05
    }
    
    # 2. KPSS 
    kpss_stat, kpss_p, _, _ = kpss(series, regression='c', nlags='auto')
    resultados['KPSS'] = {
        'Estadístico': kpss_stat,
        'p-value': kpss_p,
        'Estacionariedad (0.05)': kpss_p > 0.05
    }
    
    return pd.DataFrame(resultados).T
```

```python
def evaluate_volatility_models(returns, max_arch_p=4):
    warnings.simplefilter('ignore', FutureWarning)
    results = []
    
    COL_MODEL = 'Modelo'
    COL_SIG = 'Todos los Parametros Significativos (<0.05)'
    COL_ARCH_P = 'ARCH-LM P-value'
    COL_HOMO = 'Homocedastica (>0.05)'
    COL_PVALS = 'P-values de los Parametros del Modelo'
    
    # 1. Bucle ARCH
    for p in range(1, max_arch_p + 1):
        model = arch_model(returns, vol='ARCH', p=p, rescale=False)
        res = model.fit(disp='off')
        
        pvals = res.pvalues
        all_sig = bool(all(pvals < 0.05))
        pvals_str = ", ".join([f"{idx}: {val:.4f}" for idx, val in pvals.items()])
        
        std_resid = res.std_resid.dropna()
        _, arch_lm_pval, _, _ = het_arch(std_resid, nlags=5)
        lm_passed = bool(arch_lm_pval > 0.05)
        
        results.append({
            COL_MODEL: f'ARCH({p})',
            COL_SIG: all_sig,
            COL_ARCH_P: round(arch_lm_pval, 4),
            COL_HOMO: lm_passed,
            COL_PVALS: pvals_str
        })
        
    # 2. Bucle GARCH
    garch_configs = [(1, 1), (1, 2), (2, 1), (2, 2)]
    trigger_egarch = False 
    
    for p, q in garch_configs:
        model = arch_model(returns, vol='GARCH', p=p, q=q, rescale=False, dist="normal", mean="Constant")
        res = model.fit(disp='off')
        
        pvals = res.pvalues
        all_sig = bool(all(pvals < 0.05))
        pvals_str = ", ".join([f"{idx}: {val:.4f}" for idx, val in pvals.items()])
        
        if 'mu' in pvals:
            mu_no_sig = bool(pvals['mu'] >= 0.05)
            resto_sig = bool(all(pvals.drop('mu') < 0.05))
            
            if mu_no_sig and resto_sig:
                trigger_egarch = True
        
        std_resid = res.std_resid.dropna()
        _, arch_lm_pval, _, _ = het_arch(std_resid, nlags=5)
        lm_passed = bool(arch_lm_pval > 0.05)
        
        results.append({
            COL_MODEL: f'GARCH({p},{q})',
            COL_SIG: all_sig,
            COL_ARCH_P: round(arch_lm_pval, 4),
            COL_HOMO: lm_passed,
            COL_PVALS: pvals_str
        })
        
    # 3. Bucle EGARCH
    if trigger_egarch:
        for p, q in garch_configs:
            try:
                model = arch_model(returns, vol='EGARCH', p=p, q=q, rescale=False, dist="normal", mean="Constant")
                res = model.fit(disp='off')
                
                pvals = res.pvalues
                all_sig = bool(all(pvals < 0.05))
                pvals_str = ", ".join([f"{idx}: {val:.4f}" for idx, val in pvals.items()])
                
                std_resid = res.std_resid.dropna()
                _, arch_lm_pval, _, _ = het_arch(std_resid, nlags=5)
                lm_passed = bool(arch_lm_pval > 0.05)
                
                results.append({
                    COL_MODEL: f'EGARCH({p},{q})',
                    COL_SIG: all_sig,
                    COL_ARCH_P: round(arch_lm_pval, 4),
                    COL_HOMO: lm_passed,
                    COL_PVALS: pvals_str
                })
            except Exception as e:
                pass
            
    return pd.DataFrame(results)
```

Al no descubrir ninguna estacionalidad, quitaremos solamente la tendencia y observaremos la homocedasticidad.
```python
CAT_r = CAT.pct_change().dropna()*100
```

```python
CAT_r.plot()
```

```python
test_estacionariedad(CAT_r)
```

Por lo tanto, estacionaria en media.

Ahora se trata la varianza.

```python
CAT_sq = CAT_r ** 2

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
tsaplots.plot_acf(CAT_sq, lags=20, ax=ax[0])
tsaplots.plot_pacf(CAT_sq, lags=20, ax=ax[1])
plt.show()
```

Una clara falta de variación geométrica con algun regazo significativo a considerar.

Por lo tanto, dejamos ARCH de lado y pasamos con GARCH y EGARCH.

```python
tabla_resumen = evaluate_volatility_models(CAT_r, max_arch_p=0)
print(tabla_resumen.to_string(index=False))
```

GARCH no significativo por excepción de mu -> Prueba de EGARCH -> EGARCH no significativo -> Volver al GARCH y asumir la no significatividad en media


Asumiendo sus valores del análisis de PACF y ACF, el GARCH(1,1) se muestra ganador a pesar del mu no significativo, modelo con el que se predicen las volatilidades
para los próximos 22 días.

```python
model_garch = arch_model(CAT_r, vol='Garch', p=1, q=1, rescale=False, dist='normal', mean='Constant')
res_garch = model_garch.fit(disp='off')

horizonte = 22
forecasts = res_garch.forecast(horizon=horizonte, reindex=False)
pred_vol = np.sqrt(forecasts.variance.iloc[-1].values)

hist_vol = res_garch.conditional_volatility[-100:] 

ultima_fecha = CAT_r.index[-1]
fechas_futuras = pd.date_range(start=ultima_fecha + pd.Timedelta(days=1), periods=horizonte, freq='B')
serie_prediccion = pd.Series(pred_vol, index=fechas_futuras)

serie_prediccion.loc[ultima_fecha] = hist_vol.iloc[-1]
serie_prediccion = serie_prediccion.sort_index()

fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(hist_vol.index, hist_vol.values, 
        color=paleta_bbva["azul_logotipo"], linewidth=2, 
        label='Volatilidad Condicional Histórica (Estimada)')

ax.plot(serie_prediccion.index, serie_prediccion.values, 
        color=paleta_bbva["naranja_alerta"], linewidth=2.5, linestyle='--', 
        label='Volatilidad Predicha GARCH(1,1) ($t+22$)')


ax.set_title("Proyección Continua de Volatilidad: Histórico vs. Predicción a 1 Mes (CAT)", 
             fontsize=14, fontweight='bold')
ax.set_ylabel("Volatilidad Diaria")
ax.set_xlabel("Fecha")
ax.legend(facecolor=paleta_bbva["blanco"], edgecolor=paleta_bbva["gris_bordes"])

plt.savefig("../DATOS/GRAFICOS/figura_prediccion_volatilidad_CAT.png", dpi=300, bbox_inches='tight')
plt.show()
```
